## 1. Introduction
**This work is incomplete.** It is an attempt at a semi-supervised learning model (SSL) combining the supervised K Nearest Neighbour (KNN) and unsupervised  K-Means models.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
df = pd.read_csv("sdss_star_classification.csv")
```

```python
features = ['u', 'g', 'r', 'i', 'z', 'alpha', 'delta']
target = 'redshift'
```

splitting data into labelled and unlabelled

```python
labeled_data = df[df[target].notnull()].copy()
unlabeled_data = df[df[target].isnull()].copy()
```

```python
scaler = StandardScaler()
labeled_features_scaled = scaler.fit_transform(labeled_data[features])
```

## K MEANS

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
labeled_data['cluster'] = kmeans.fit_predict(labeled_features_scaled)
```

## KNN

```python
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(labeled_features_scaled, labeled_data[target])
```

silhouette score

```python
from sklearn.metrics import silhouette_score, mean_absolute_error, mean_squared_error
```

MAE and MSE for KNN

```python
knn_predictions = knn.predict(labeled_features_scaled)
mae = mean_absolute_error(labeled_data[target], knn_predictions)
mse = mean_squared_error(labeled_data[target], knn_predictions)
print(f"KNN Mean Absolute Error (MAE): {mae:.6f}")
print(f"KNN Mean Squared Error (MSE): {mse:.6f}")
```
